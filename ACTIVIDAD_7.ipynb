```python
# ============================================================
# PUNTO 2 - TODO EN UNA SOLA CELDA (Google Colab)
# ============================================================

# ---------- PASO 1: Instalar y arrancar MySQL ----------
print("⏳ Instalando MySQL... (puede tardar 1 minuto)")
!apt-get update -qq
!apt-get install mysql-server -y -qq
!service mysql start
!pip install mysql-connector-python -q
print("✅ MySQL instalado y corriendo\n")

# ---------- PASO 2: Poner contraseña a root ----------
!mysql -u root -e "ALTER USER 'root'@'localhost' IDENTIFIED WITH mysql_native_password BY 'colab_pass';"
print("✅ Contraseña de root configurada\n")

# ---------- PASO 3: Conectar y crear BD + tabla ----------
import mysql.connector

cnx = mysql.connector.connect(
    host="127.0.0.1",
    user="root",
    password="colab_pass",
    charset="utf8mb4"
)
cursor = cnx.cursor(buffered=True)

cursor.execute("CREATE DATABASE IF NOT EXISTS taller_python")
cursor.execute("USE taller_python")

cursor.execute("""
    CREATE TABLE IF NOT EXISTS estudiantes (
        id INT AUTO_INCREMENT PRIMARY KEY,
        nombre VARCHAR(100) NOT NULL,
        edad INT,
        correo VARCHAR(100)
    )
""")
print("✅ Base de datos 'taller_python' y tabla 'estudiantes' listas\n")

# ---------- PASO 4: Insertar 3 registros ----------
estudiantes = [
    ("Ana Gómez", 20, "ana@email.com"),
    ("Carlos Ruiz", 22, "carlos@email.com"),
    ("María López", 21, "maria@email.com")
]

cursor.executemany(
    "INSERT INTO estudiantes (nombre, edad, correo) VALUES (%s, %s, %s)",
    estudiantes
)
cnx.commit()
print(f"✅ {cursor.rowcount} registros insertados\n")

# ---------- PASO 5: Consultar y mostrar ----------
cursor.execute("SELECT id, nombre, edad, correo FROM estudiantes")

print("📋 ESTUDIANTES REGISTRADOS")
print("-" * 60)
for fila in cursor:
    print(f"ID: {fila[0]} | Nombre: {fila[1]} | Edad: {fila[2]} | Correo: {fila[3]}")

# ---------- PASO 6: Cerrar conexión ----------
cursor.close()
cnx.close()
print("\n🔒 Conexión cerrada correctamente")
print("\n🎉 ¡Punto 2 completado!")
```
